Safe Division Utility

```python
try:

    numerator = int(input("Enter the numerator: "))
    denominator = int(input("Enter the denominator: "))

    result = numerator / denominator

except ZeroDivisionError:
    print("Error: Division by zero is not allowed.")
except ValueError:
    print("Error: Invalid input. Please enter valid integers.")

else:
    print(f"The result of {numerator} divided by {denominator} is {result}")
finally:
    print("Division operation complete.")
```

Bill Calculator with Error Handling 

```python
prices = [120, 350, 'abc', 500, -200, 800]
total = 0
for price in prices:
    try:
        if not isinstance(price, (int, float)):
            raise TypeError("Value is not a number")
        if price < 0:
            raise ValueError("Negative price not allowed")
        total += price
    except ValueError as e:
        print(f"Error: {e}")
    except TypeError as e:
        print(f"Error: {e}")
    print(f"Running total: {total}")

print(f"Total bill amount: {total}")
```

Custom Exception : Age Validator

```python
def check_age(age):
    if age < 1 or age > 120:
        raise ValueError("Age must be between 1 and 120")


age = int(input("Enter your age: "))
try:
    check_age(age)
    print("Age is valid")
except ValueError as e:
    print(f"Error: {e}")
        
```

```python
# Custom exception for negative prices
class NegativePriceError(Exception):
    """Raised when a negative price is entered."""
    pass


cart = []
while True:
    user_input = input("Enter item price (or 'q' to quit): ").strip()
    if user_input.lower() == 'q':
        break
    try:
        price = float(user_input)
        if price < 0:
            raise NegativePriceError("Negative price not allowed.")
        cart.append(price)
    except ValueError:
        print("Error: Invalid input. Please enter a valid number.")
    except NegativePriceError as e:
        print(f"Error: {e}")

print(f"Total items: {len(cart)}")
print(f"Total bill: {sum(cart):.2f}")
```
