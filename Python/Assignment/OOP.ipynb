Basic Class and Object Creation

```python
class Product:
    def __init__(self,name,price,category):
        self.name = name
        self.price = price
        self.category = category

    def get_info(self):
        return f"Product: {self.name}, Price: {self.price}, Category: {self.category}"
    
    def apply_discount(self,discount):
        self.price = self.price * (1-(discount/100))
        return self.price
    
    def __str__(self):
        return f"Product: {self.name}, Price: {self.price}, Category: {self.category}"
    
```

```python
Car = Product("Toyota",10000,"Car")
print(Car.get_info())
Soap = Product("Dove",10,"Soap")
print(Soap.get_info())
Car.apply_discount(10)
print(Car.get_info())
```

Constructor and Encapsulation

```python
class Product:
    def __init__(self,name,price,category):
        self.name = name
        self.__price = price
        self.category = category

    def get_price(self):
        return self.__price
    
    def set_price(self,new_price):
        if new_price > 0:
            self.__price = new_price
            print(f"Price of {self.name} updated to {new_price}")
        else:
            print("Price must be positive")
```

```python
p1 = Product("Laptop",1000,"Electronics")
p1.set_price(1500)
p1.set_price(-500)
```

Inheritance (Single Level)

```python
class Product:
    def __init__(self, name, price, category):
        self.name = name
        self.__price = price
        self.category = category

    def get_price(self):
        return self.__price

    def set_price(self, new_price):
        if new_price > 0:
            self.__price = new_price
            print(f"Price of {self.name} updated to {new_price}")
        else:
            print("Price must be positive")

    def get_info(self):
        return f"Product: {self.name}, Price: {self.__price}, Category: {self.category}"

class ElectronicProduct(Product):
    def __init__(self,name,price,category,warranty_period):
        super().__init__(name,price,category)
        self.warranty_period = warranty_period

    def get_info(self):
        return f"{super().get_info()}, Warranty: {self.warranty_period} months"
```

```python
ep1 = ElectronicProduct("Laptop",1000,"Electronics",12)
print(ep1.get_info())

ep1.set_price(1500)
print(ep1.get_info())
```

Polymorphism

```python
class Laptop(Product):
    def __init__(self,name,price,category,ram):
        super().__init__(name,price,category)
        self.ram = ram

    def get_info(self):
        return f"{super().get_info()}, RAM: {self.ram} GB"

class Mobile(Product):
    def __init__(self,name,price,category,camera_resolution):
        super().__init__(name,price,category)
        self.camera_resolution = camera_resolution

    def get_info(self):
        return f"{super().get_info()}, Camera Resolution: {self.camera_resolution} MP"

items = [
    Laptop("MacBook Pro",150000,"Electronics",16),
    Mobile("iPhone 15",100000,"Electronics",12),
    Laptop("Dell Inspiron",80000,"Electronics",8),
    Mobile("Samsung Galaxy S22",70000,"Electronics",10)
]

for item in items:
    print(item.get_info())
```

```python
from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def process_payment(self,amount):
        pass

class CreditCardPayment(Payment):
    def process_payment(self,amount):
        print(f"Processing credit card payment of {amount} USD")

class UPIPayment(Payment):
    def process_payment(self,amount):
        print(f"Processing UPI payment of {amount} USD")
```

```python
p1 = CreditCardPayment()
p1.process_payment(1000)
p2 = UPIPayment()
p2.process_payment(1000000)
```

Magic Methods and Operator Overloading

```python
class Product:
    def __init__(self,name,price,category):
        self.name = name
        self.price = price
        self.category = category

    def __str__(self):
        return f"Product({self.name}, {self.price}, {self.category})"

    def __add__(self,other):
        return f"Total Price: {self.price+other.price} , All Products: {self.name + " & " + other.name}"
```

```python
p1=Product("Laptop",1000,"Electronics")
print(p1)

p2=Product("Mobile",500,"Electronics")
print(p2)

team = p1+p2
print(team)
```

Mini Project : Simple Inventory System

```python
class Product:
    def __init__(self, name, price, category):
        self.name = name
        self.price = price
        self.category = category

    def __str__(self):
        return f"Product: {self.name}, Price: {self.price}, Category: {self.category}"

    def __add__(self, other):
        return self.price + other.price


class Inventory:
    def __init__(self):
        self.products = []

    def add_product(self, product):
        self.products.append(product)

    def remove_product(self, name):
        self.products = [product for product in self.products if product.name != name]

    def get_total_value(self):
        return sum(product.price for product in self.products)

    def show_all_products(self):
        for product in self.products:
            print(product)

class Store:
    def __init__(self, store_name):
        self.store_name = store_name
        self.inventory = Inventory()

    def add_new_product(self, name, price, category):
        name = input("Enter product name: ")
        price = float(input("Enter product price: "))
        category = input("Enter product category: ")

        product = Product(name, price, category)
        self.inventory.add_product(product)

    def show_summary(self):
        total_items = len(self.inventory.products)
        total_value = self.inventory.get_total_value()
        print(f"Store: {self.store_name}")
        print(f"Total Items: {total_items}")
        print(f"Total Inventory Value: {total_value}")
```

```python
my_store = Store("TechMart")

my_store.add_new_product("Laptop",1000,"Electronics")
my_store.add_new_product("Mobile",500,"Electronics")
my_store.add_new_product("Tablet",300,"Electronics")

my_store.show_summary()

my_store.inventory.show_all_products()

my_store.inventory.remove_product("Laptop")
my_store.inventory.show_all_products()

my_store.inventory.get_total_value()


```

```python
p1 = my_store.inventory.products[0]
print(p1)
p2 = my_store.inventory.products[1]
print(p2)

print(f"Combined Price: {p1+p2} for {p1.name} and {p2.name}")
```
