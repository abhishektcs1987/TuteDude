Basic Function - Price after Discount

```python
def apply_discount(price, discount_rate=5):
    if discount_rate == 0:
        return price - (price * discount_rate/100)

    return price - (price * discount_rate/100)
```

```python
apply_discount(1000 , 10)
```

```python
apply_discount(500)
```

Factorial

```python
def factorial(n):
    if n < 0:
        return "Factorial is not defined for negative numbers"
    if n==0:
        return 1
    if n == 1:
        return 1
    else:
        return n * factorial(n-1)
```

```python
factorial(5)
```

```python
factorial(0)
```

```python
factorial(-3)
```
