Basic Function - Price after Discount

```python
def apply_discount(price, discount_rate=5):
    if discount_rate == 0:
        return price - (price * discount_rate/100)

    return price - (price * discount_rate/100)
```

```python
apply_discount(1000 , 10)
```

```python
apply_discount(500)
```

Factorial

```python
def factorial(n):
    if n < 0:
        return "Factorial is not defined for negative numbers"
    if n==0:
        return 1
    if n == 1:
        return 1
    else:
        return n * factorial(n-1)
```

```python
factorial(5)
```

```python
factorial(0)
```

```python
factorial(-3)
```

Lambda Function:GST Calculator

```python
def gst_calculator(price):
    gst = lambda price: price + (price * 0.18)
    return gst(price)
```

```python
print(gst_calculator(100))
```

```python
final_price = lambda price,discount_rate: price - (price * discount_rate/100) * 1.18

final_price(100,10)
```

Using map()

```python
prices = [100,250,400,1200,50]

prices_with_gst = list(map(lambda price: price + (price * 0.18), prices))
print("Original prices:", prices)
print("Prices with GST:", prices_with_gst)

```
