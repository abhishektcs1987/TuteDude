Basic Function - Price after Discount

```python
def apply_discount(price, discount_rate=5):
    if discount_rate == 0:
        return price - (price * discount_rate/100)

    return price - (price * discount_rate/100)
```

```python
apply_discount(1000 , 10)
```

```python
apply_discount(500)
```

Factorial

```python
def factorial(n):
    if n < 0:
        return "Factorial is not defined for negative numbers"
    if n==0:
        return 1
    if n == 1:
        return 1
    else:
        return n * factorial(n-1)
```

```python
factorial(5)
```

```python
factorial(0)
```

```python
factorial(-3)
```

Lambda Function:GST Calculator

```python
def gst_calculator(price):
    gst = lambda price: price + (price * 0.18)
    return gst(price)
```

```python
print(gst_calculator(100))
```

```python
final_price = lambda price,discount_rate: price - (price * discount_rate/100) * 1.18

final_price(100,10)
```

Using map()

```python
prices = [100,250,400,1200,50]

prices_with_gst = list(map(lambda price: price + (price * 0.18), prices))
print("Original prices:", prices)
print("Prices with GST:", prices_with_gst)

```

Using filter(): Filter Expensive Products

```python
prices = [100,250,400,1200,50,2000,850]

expensive_products = list(filter(lambda price: price >500, prices))
cheap_products = list(filter(lambda price: price <=500, prices))
print("Expensive products:", expensive_products)
print("Cheap products:", cheap_products)
```

Combined Utility Function

```python
def process_prices(prices):
    discounted_prices = list(map(lambda price: price - (price * 0.1), prices))
    filtered_prices = list(filter(lambda price: price > 300, discounted_prices))
    return discounted_prices, filtered_prices

process_prices([100,500,900,50,750])

```

Mini Problem : Menu Using Functions

```python
def add_price(prices_list, price):
    prices_list.append(price)


def get_average_price(prices_list):
    if not prices_list:
        return None
    return sum(prices_list) / len(prices_list)


def get_max_price(prices_list):
    if not prices_list:
        return None
    return max(prices_list)


prices_list = []

while True:
    print("1 -> Add price")
    print("2 -> Show average price")
    print("3 -> Show highest price")
    print("q -> Quit")

    choice = input("Enter your choice: ").strip()

    if choice == "1":
        try:
            price = float(input("Enter price: "))
        except ValueError:
            print("Invalid price. Please enter a number.")
            continue
        add_price(prices_list, price)
        print("Price added.")

    elif choice == "2":
        avg = get_average_price(prices_list)
        if avg is None:
            print("No prices available.")
        else:
            print(f"Average price: {avg:.2f}")

    elif choice == "3":
        max_price = get_max_price(prices_list)
        if max_price is None:
            print("No prices available.")
        else:
            print(f"Highest price: {max_price:.2f}")

    elif choice == "q":
        break
    
    else:
        print("Invalid choice. Please try again.")
```
