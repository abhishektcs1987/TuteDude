Product Collections (Lists & Tuples)

```python
#list 
products=["Laptop","Mouse","Keyboard","Monitor","Speaker","Tablet"]
```

```python
#tuple named sample_product that stores (product_name,price,category)   
sample_product=("Laptop",50000,"Electronics")
```

```python
#Print the second and last product from the list
print(products[1])
print(products[-1])
```

```python
products.append("Headphones")
products.append("Charger")
print(products)
```

```python
#Convert sample_product to a list
sample_product_list=list(sample_product)
print(sample_product_list)
sample_product_list[1]=70000
print(sample_product_list)
sample_product_tuple=tuple(sample_product_list)
print(sample_product_tuple)
print(type(sample_product_tuple))
```

Categories (Sets)

```python
categories=["Laptop","Accessories","Accessories","Electronics","Accessories","Electronics","Audio","Charger"]
```

```python
categories_set=set(categories)
print(categories_set)
```

```python
#Duplicates Ignored
categories_set.add("Phone")
categories_set.add("Audio")
print(categories_set)
```

```python
print("Is 'Electronics' in categories_set?", "Electronics" in categories_set)
```

```python
#total number of unique categories using a set
print("Total unique categories:", len(categories_set))
```
