Dictionaries

```python
#Create price_dict
price_dict = {
    "Laptop": 50000,
    "Mouse": 1000,
    "Keyboard": 2000,
    "Monitor": 30000,
    "Speaker": 5000,
    "Tablet": 15000,
}
```

```python
#Add a new product
price_dict["Headphones"]=1000
```

```python
#Update an existing prodcut price
price_dict['Mouse']=1500
```

```python
product_to_remove="Abhishek"
if product_to_remove in price_dict:
    del price_dict[product_to_remove]
    print(f"{product_to_remove} removed from price_dict")
else:
    print(f"{product_to_remove} not found in price_dict")
    
```

```python
#
total_price=sum(price_dict.values())
average_price=total_price/len(price_dict)
print(f"Average price of all products: {average_price:.2f}")
```

```python
#Max Price and Min Price
max_price=max(price_dict.values())
min_price=min(price_dict.values())
print(f"The highest price is {max_price} and the lowest price is {min_price}")
#Max Product and Min Product
max_product=max(price_dict,key=price_dict.get)
min_product=min(price_dict,key=price_dict.get)
print(f"The most expensive product is {max_product} with a price of {price_dict[max_product]}")
print(f"The cheapest product is {min_product} with a price of {price_dict[min_product]}")
```

Combined Operations

```python
#Create Catalog
catalog = [(prod, price_dict[prod], cat) for prod, cat in zip(products, categories)]
print("Catalog (product, price):", catalog)
```

```python
type(catalog)
```

```python
#category -> product names
category_to_products = {}
for prod,price,cat in catalog:
    category_to_products.setdefault(cat,[]).append((prod,price))

print("Category map:", category_to_products)
```

```python
#Maximum Products in a Category
max_category = max(category_to_products,key=lambda c:len(category_to_products[c]))
print(f"The category with the most products is {max_category} with {len(category_to_products[max_category])} products")
print("Products:", category_to_products[max_category])
```
