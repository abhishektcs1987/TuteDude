Write Sales Records to a File

```python
sales = [1200,450,980,1500,3000]
f=open("sales_data.txt", "w")
for sale in sales:
    f.write(f"{sale}\n")
f.close()
with open("sales_data.txt", "r") as f:
    for line in f:
        print(line.strip())
```

Append New Sales

```python
with open("sales_data.txt", "a") as f:
    f.write("5000\n")
    f.write("2500\n")
    f.write("1700\n")
f.close()
with open("sales_data.txt", "r") as f:
    for line in f:
        print(line.strip())
```

```python
count=0
with open("sales_data.txt", "r") as f:
    for line in f:
        count += 1
    print(count)
```

Generate Summary Report from File

```python
with open("sales_data.txt", "r") as f:
    lines=f.readlines()
    integer_list = [int(line.strip()) for line in lines]
    total_sales = sum(integer_list)
    print(f"Total Sales: {total_sales}")
    highest_sale = max(integer_list)
    print(f"Highest Sale: {highest_sale}")
    lowest_sale = min(integer_list)
    print(f"Lowest Sale: {lowest_sale}")
    average_sale = total_sales / count
    print(f"Average Sale: {average_sale}")
```

Read file Safely

```python
import os

filename = input("Enter filename to open: ")

if os.path.exists(filename):
    with open(filename, "r") as f:
        print(f.read())
else:
    print(f"File {filename} does not exist.")
```

Mini Project

```python
prices = {
    "Mouse": 500,
    "Keyboard": 800,
    "Monitor": 7000,
    "Pendrive": 400,
    "Camera": 5000
}

discount_percent = float(input("Enter discount percentage: "))

with open("discounted_report.txt", "w") as f:
    f.write("Product Name | Original Price | Discounted Price\n")

stotal_discounted = 0
count = 0

for product, price in prices.items():
    discounted_price = price - (price * discount_percent / 100)
    total_discounted += discounted_price
    count += 1

 # Extra summary
    average_discounted = total_discounted / count
    f.write(f"\nTotal Items: {count}\n")
    f.write(f"Average Discounted Price: {average_discounted:.2f}\n")

# Read and print the report
with open("discount_report.txt", "r") as f:
    print(f.read())
    
```
