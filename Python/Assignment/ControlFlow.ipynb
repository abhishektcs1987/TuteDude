```python
# Task 1: Discount Rules
try:
    order_amount = float(input("Order Amount: "))
except ValueError:
    print("Error: Please enter a valid number for order amount.")
    raise SystemExit

if order_amount >= 2000:
    discount_rate = 0.15
elif order_amount >= 1500:
    discount_rate = 0.10
elif order_amount >= 1000:
    discount_rate = 0.07
else:
    discount_rate = 0.0

discount = order_amount * discount_rate
subtotal = order_amount - discount

tax_rate = 0.05
tax = subtotal * tax_rate
final_total = subtotal + tax

print(f"Subtotal after discount: {subtotal:.2f}")
print(f"Tax (5%): {tax:.2f}")
print(f"Final total: {final_total:.2f}")
```

```python
orders = [1200, 2500, 800, 1750, 3000]

total_revenue = 0
discounted_count = 0

print("order_amount -> discount% -> final_amount")

for order_amount in orders:
    if order_amount >= 2000:
        discount_rate = 0.15
    elif order_amount >= 1500:
        discount_rate = 0.10
    elif order_amount >= 1000:
        discount_rate = 0.07
    else:
        discount_rate = 0.0

    discount = order_amount * discount_rate
    final_amount = order_amount - discount
    total_revenue += final_amount

    if discount_rate > 0:
        discounted_count += 1

    print(f"{order_amount} -> {discount_rate*100:.0f}% -> {final_amount:.2f}")

print("Total revenue after discounts:", round(total_revenue, 2))
print("Orders with discount:", discounted_count)


```

```python
choice = 1
orders = []
while(choice):
    print("1. Add order amount to a running list")
    print("2. Show all orders and total after applying discounts")
    print("q. Quit")

    choice = input("Enter your choice: ")

    if choice == "1":
        try:
            order_amount = float(input("Order Amount: "))
        except ValueError:
            print("Error: Please enter a valid number for order amount.")
            raise SystemExit
        orders.append(order_amount)

    elif choice == "2":
        total_revenue = 0
        discounted_count = 0

        for order_amount in orders:
            if order_amount >= 2000:
                discount_rate = 0.15
            elif order_amount >= 1500:
                discount_rate = 0.10
            elif order_amount >= 1000:
                discount_rate = 0.07
            else:
                discount_rate = 0.0

        total_revenue += order_amount - (order_amount * discount_rate)
        print(f"Total revenue after discounts: {total_revenue:.2f}")
        print("All orders:")
        for order in orders:
            print(order)

    elif choice == "q":
        break

    elif choice:
        continue
    
```

Loop Control with Conditions

```python
daily=[200,150,0,400,50,-1,300]
total_sales=0
for i in daily:
    if i == -1:
        print("Corrupted data found. Stopping.")
        break

    elif i == 0:
        continue

    else:
        total_sales += i
        print(f"Adding {i} to total sales. Total sales: {total_sales}")

print(f"Total sales: {total_sales}")
```
